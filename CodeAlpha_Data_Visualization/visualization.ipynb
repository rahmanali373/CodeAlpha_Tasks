# CodeAlpha: Task 3 - Data Visualization
**Objective:** Transform raw e-commerce data into compelling visual formats (charts and graphs) using Python, Matplotlib, and Seaborn to reveal clear business insights and support decision-making.

```python
# 1. Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load the e-commerce dataset
df = pd.read_csv('Amazon Sale Report.csv', low_memory=False)
# 3. Perform data cleaning (handling missing values and dropping redundant columns)
df.drop(columns=['Unnamed: 22', 'fulfilled-by', 'promotion-ids'], inplace=True, errors='ignore')
df.dropna(subset=['Amount'], inplace=True)

```

### 1. Correlation Heatmap
**Insight:** This matrix checks the linear relationship between numeric features like order quantities and monetary amounts, establishing that pricing strategies and quantities are managed independently per transaction.

```python

# 4. Create a Correlation Heatmap to analyze relationships between numeric features
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=['number'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

# 5. Add title to the heatmap
plt.title('Correlation Heatmap of Numeric Features', fontsize=14, fontweight='bold')
plt.show()
```

### 2. Price Distribution by Product Category (Box Plot)
**Insight:** This visualization highlights price variations and outlier pricing across different product categories, helping stakeholders understand spending patterns and premium product segments.


```python
# 6.Create a Box Plot to show price distribution across different product categories
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Category', y='Amount', palette='Set2')

# 7. Add titles and labels to the box plot
plt.title('Price Distribution by Product Category (Box Plot)', fontsize=14, fontweight='bold')
plt.xlabel('Category', fontsize=12)
plt.ylabel('Amount (INR)', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### 3. Top 10 States by Total Sales (Bar Chart)
**Insight:** This visualization identifies the top-performing geographic regions. Understanding which states generate the most revenue allows for targeted marketing and optimized supply chain logistics in those specific areas.

```python
# 8. Group data by state to find the top 10 states with the highest total sales
plt.figure(figsize=(12, 6))
top_states = df.groupby('ship-state')['Amount'].sum().sort_values(ascending=False).head(10)

# 9. Create a Bar Chart
sns.barplot(x=top_states.index, y=top_states.values, palette='viridis')

# 10. Add titles and labels
plt.title('Top 10 States by Total Sales', fontsize=14, fontweight='bold')
plt.xlabel('State', fontsize=12)
plt.ylabel('Total Sales (INR)', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### 4. Order Status Distribution (Donut Chart)
**Insight:** This chart provides a clear breakdown of order fulfillment statuses (e.g., Shipped, Cancelled). Tracking these proportions is crucial for evaluating operational efficiency and customer satisfaction.

```python
# 11. Count the occurrences of the top 5 order statuses
plt.figure(figsize=(8, 8))
status_counts = df['Status'].value_counts().head(5)

# 12. Create a Pie Chart
plt.pie(status_counts, labels=status_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))

# 13. Draw a white circle in the center to turn the Pie Chart into a Donut Chart
centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

# 14. Add title
plt.title('Top 5 Order Status Distribution', fontsize=14, fontweight='bold')
plt.show()
```
